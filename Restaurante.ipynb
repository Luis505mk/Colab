```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

Creamos la tabla llamada **comidas**


```python
cur.executescript('''
DROP TABLE IF EXISTS comidas;

CREATE TABLE comidas (
  id_Platillo INTEGER PRIMARY KEY,
  nombre varchar(50)  NOT NULL,
  categoria varchar(30),
  precio decimal(6,2),
  disponible boolean
);
''')
conn.commit()
print("Tabla creadas")
```

Realizamos dos inser uno de **Refresco** y **sopa**

```python
#insert:de refresco
run("""
INSERT INTO comidas (nombre, categoria, precio, disponible)
VALUES ('Refresco', 'Bebida', 25.00, 1)
""")

conn.commit()
```

```python
#insert:de comida

run(""" INSERT INTO comidas ( nombre, categoria, precio ) VALUES

('Sopa', 'entrada', 100 );""")

query("select * from comidas")
```

Realizamos la actualizacion del precio de la **Sopa** con un **UPDATE**



```python
#update

cur.execute("""
UPDATE comidas
SET precio = 120
WHERE nombre = 'Sopa'
""")

conn.commit()

print("Precio actualizado")
query("SELECT * FROM comidas")
```

Eliminamos un platillo con el **id 1**, correspondiente al **Refresco** y mostramos la tabla nuevamente

```python
#DELETE
cur.execute("""
DELETE FROM comidas
WHERE id_Platillo = 1
""")

conn.commit()

query("SELECT * FROM comidas")
```
