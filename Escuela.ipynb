```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

Creamos la tabla llamada **alumnos**

```python
#Creamos la tabla alumnos
cur.executescript('''
DROP TABLE IF EXISTS alumnos;

CREATE TABLE alumnos (
    id_alumno INTEGER PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    grado VARCHAR(10),
    edad INTEGER,
    promedio DECIMAL(4,2)
);
''')

conn.commit()
print("Tabla alumnos creada")
```

Realizamos varios insert de **alumnos** y mostramos la tabla

```python
#INSERT
run("""
INSERT INTO alumnos
VALUES (1, 'Juan Perez', '3A', 15, 8.5)
""")

run("""
INSERT INTO alumnos
VALUES (2, 'Ana Lopez', '5B', 17, 9.2)
""")

run("""
INSERT INTO alumnos
VALUES (3, 'Carlos Ruiz', '4A', 16, 8.8)
""")

conn.commit()

query("SELECT * FROM alumnos")
```

Realizamos un **UPDATE** al promedio de **Juan Perez** de **8.5** a **9.8**

```python
#UPDATE
cur.execute("""
UPDATE alumnos
SET promedio = 9.8
WHERE nombre = 'Juan Perez'
""")

conn.commit()

query("SELECT * FROM alumnos")
```

Eliminamos de la tabla a **Carlos Ruiz** y mostramos la tabla

```python
#DELETE
cur.execute("""
DELETE FROM alumnos
WHERE nombre = 'Carlos Ruiz'
""")

conn.commit()

query("SELECT * FROM alumnos")
```
