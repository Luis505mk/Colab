```python
import sqlite3
import pandas as pd

# Conexión en memoria (se pierde al cerrar el notebook)
conn = sqlite3.connect(":memory:")
cur = conn.cursor() #objeto para ejecutar comandos de SQL

def query(sql, params=()):
    """Ejecuta un SELECT y devuelve un DataFrame para verlo bonito."""
    return pd.read_sql_query(sql, conn, params=params)

def run(sql, params=()):
    """Ejecuta INSERT/UPDATE/DELETE/DDL y hace commit."""
    cur.execute(sql, params)
    conn.commit()
    print("OK ->", sql.strip().splitlines()[0], "...")
```

```python
#Creamos la tabla productos
cur.executescript('''
DROP TABLE IF EXISTS productos;

CREATE TABLE productos (
    id_producto INTEGER PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    tipo VARCHAR(30),
    precio DECIMAL(6,2),
    existencia INTEGER
);
''')

conn.commit()
print("Tabla productos creada")
```

```python
#INSERT

run("""
INSERT INTO productos
VALUES (1, 'Papas', 'Snack', 18.50, 20)
""")

run("""
INSERT INTO productos
VALUES (2, 'Agua', 'Bebida', 12.00, 30)
""")

run("""
INSERT INTO productos
VALUES (3, 'Galletas', 'Snack', 15.00, 25)
""")

conn.commit()

query("SELECT * FROM productos")
```

```python
#UPDATE
cur.execute("""
UPDATE productos
SET existencia = 50
WHERE nombre = 'Papas'
""")

conn.commit()

query("SELECT * FROM productos")
```

```python
#DELETE
cur.execute("""
DELETE FROM productos
WHERE nombre = 'Galletas'
""")

conn.commit()

query("SELECT * FROM productos")
```
